<!-- RunnableLambda Code -->

```python
from langchain_core.runnables import RunnableLambda
r=RunnableLambda(lambda x:x+1)
result=r.invoke(10)
print(result)
```

<!-- RunnableSequence -->

```python
# it connects multi runnables
Sequence=(RunnableLambda(lambda x:x+1) | RunnableLambda(lambda x:x*2))
print(Sequence.invoke(2))
```

<!-- RunnableParallel -->

```python
# it retuens in dict and take same input.
from langchain_core.runnables import RunnableParallel
rp=RunnableParallel({
    "mul_2":RunnableLambda(lambda x:x*2),
    "mul_5":RunnableLambda(lambda x:x*5)
})
result=rp.invoke(5)
print(result)
```

```python
# it pass input to output wothout modify
from langchain_core.runnables import RunnablePassthrough
rp=RunnablePassthrough()
print(rp.invoke("Hello uday"))
```

```python
# it return the only requiredd key-value pairs
from langchain_core.runnables import RunnablePick
rp = RunnablePick("name")

data={
    "name":"uday",
    "skill":"Python",
    "city":"Benglore"
}
print(rp.invoke (data))
```

```python
# it checks condition in chain processes
from langchain_core.runnables import RunnableLambda , RunnableBranch
branch=RunnableBranch(
    (lambda x:x>=35,RunnableLambda(lambda x:"Pass")),
    RunnableLambda(lambda x:"fail")
)
print(branch.invoke(60))
print(branch.invoke(20))
```

```python
from langchain_core.runnables import RunnableLambda, RunnablePassthrough
rp=RunnablePassthrough.assign(total=RunnableLambda(lambda data:data["income"]-data["expenses"]))
result = rp.invoke({
    "income": 30000,
    "expenses": 20000
})
print(result)
```
